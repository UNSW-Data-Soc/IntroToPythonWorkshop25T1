```python
# 1. Varibles in Python

# Unlike other languages, you don't need to declare the type of the variable.
# This is because it knows from what value it is assigned (dynamic typing).

johnAge = 10;
johnLastName = "Smith"

# Print function output the value in the terminal.
print(johnAge)
print(johnLastName)
print(type(johnAge))
print(type(johnLastName))
```

```python
# 3. Strings Functions

# Important to know when manipulating a string
x = "Hello World"

# String can be treated like arrays -> Collection of elements organized in a specific order.
# Hence we can print subset of a string
print(x[2:])

# Printing word in Uppercase
print(x.upper())

# Replace Letters in a word
print(x.replace("l", "y"))

# Spliting String
print(x.split(" "))

# Concatenating String -> combining multiple strings into a single larger string.
a = "One"
b = "Two"
c = a + b
print (c)
print (a + " " + b)

# Formatting String
# Use formatting strings to elegantly embed variables and control their display within text.
num = 10.1234
txt = "The value of x is: {x:.2f}"
print(f"The value of x is: {num}")
print(txt.format(x = num))

# Escape characters like `\n` and `\t` are used to control text formatting: `\n` inserts a new line, while `\t` adds a tab space for better structure and alignment.
# Escape character
# \n = new line
# \t = tab

# Some other Important ones:
x = "Hello"

# Return count of certain letter -> The code snippet counts how many times the letter `"l"` appears in the string variable `x` and prints the result.
print(x.count("l"))

# Return position of a character -> This code finds the position (index) of the first occurrence of the character "H" in the string variable x and prints its location.
print(x.index("H"))

# This code prints a multi-line string and splits it into a list of lines using the `splitlines()` method.
a = """Multiple
Line
String."""

print(a)

# Split Lines
print(a.splitlines())
```

```python
# 4.Operations

# Most common boolean Operation is
x = 1
y = 2

x == y #Checks if x is equal to y
x != y #Checks if x is not equal to y
x > y #Checks if x is greater t
x >= y

# Logic operators -> Logic operators are used to combine or negate conditions:
x and y
x or y
not y

# Others include is, in

# WARNING && / || cannot be used for and / or

# Arithmetic Operators (+ -  / *)
x % y # modulus
x ** y # power
x // y # floor division

# TIP if add back to itself use the <operator>= sign
x += y # is the equalivalent of
x = x + y
```

```python
# Loops
a = 1
b = 2

# if statement if true continue else jump to next condition
if a > b:
  print ("a is greater than b")
elif b == 0:
  print ("b is zero")
else:
  print("b is greater than a")

# while statement (precondition loop - until condition is meet)
i = 1
while i < 6:
  print(i)
  i += 1

# for statement (count loop - loop a certian number of time which is predetermine)
for x in range(2, 30, 3):
  print(x)
```

```python
# Functions

# Function are define using the method name and the variables passed into that function
def pythograsTheorem(x , y):
  return (x**2 + y**2) ** 0.5 # return

print(pythograsTheorem(3, 4))
```

```python
# Lambda

# More advanced, we can simplify simple function using lambda
x = lambda a, b : a * b
print(x(5, 6))

pythograsTheorem = lambda x, y : (x**2 + y**2) ** 0.5
print(pythograsTheorem(3, 4))

# The power of lambda is better shown when you use them as an anonymous function inside another function.
def myfunc(n):
  return lambda num : num * n

mydoubler = myfunc(2)
mydoubler = lambda num : num * 2

mytripler = myfunc(3)

print(mydoubler(11))
print(mytripler(11))
```

# Encoding our Data!
Now that we've cleaned and have an understanding of what our data is like, let's turn it into a format that is actually useable by our model!\
To do this let's use `scikit-learn`'s (abbreviated as `sklearn`) encoding functions.\
We will examine two of the most commonly used encoders:
- `sklearn.preprocessing.OrdinalEncoder`, the Ordinal Encoder which encodes to a single feature of ordinal integers.
- `sklearn.preprocessing.OneHotEncoder`, the One Hot Encoder which encodes to a number of features each of with binary integers.

But how do we know when to use each encoder? A general rule is only ever use One Hot Encoder with less than `10` unique values and Ordinal if there is already a pre-established order. But, this is highly arbitrary and only a general guideline!

```python
from sklearn.preprocessing import OneHotEncoder
from sklearn.preprocessing import OrdinalEncoder

df.head()
```

Before we even encode, how do we know which columns to encode?

```python
# Let's first find the columns we want to encode, all the categorical columns
categorical_columns = df.select_dtypes(include=['object'])
# display(categorical_columns)
print(categorical_columns.nunique().to_string())
```

```python
# We can then find all the categorical columns with low cardinality.
# print(categorical_columns.nunique() < 10)
small_cat_columns = categorical_columns.loc[:, categorical_columns.nunique() < 10]
small_cat_columns = categorical_columns[categorical_columns.columns[categorical_columns.nunique() < 10]]
# display(small_cat_columns)

# In this example let's just encode two particularly interesting columns with OHE
small_cat_columns = ['MainBranch', 'EdLevel']
```

Now let's create our actual encoder and encode our data!



```python
# Create our encoder
ohe = OneHotEncoder(sparse_output=False, drop=None, handle_unknown='ignore')

# TODO: Demonstrate fit and transform seperately
# Apply our encoder to the dataframe
ohe_encoded = ohe.fit_transform(df[small_cat_columns].fillna('Unknown'))

# Get feature names
# ohe_feature_names = [f"{col}_ENCODED_{val}" for col, vals in zip(['Category', 'Type'], ohe.categories_) for val in vals]
ohe_feature_names = ohe.get_feature_names_out(small_cat_columns)

# Create output df and concatenate to original df
ohe_df = pd.DataFrame(ohe_encoded, columns=ohe_feature_names, index=df.index)
df = pd.concat([df, ohe_df], axis=1)
# df.drop(columns=small_cat_columns, inplace=True)
```

And, repeat for ordinal encoding!

```python
# By exclusion (and visual examination) let's say for example we want to encode
# the following column with relatively high cardinality.
large_cat_columns = ['DevType']
```

```python
# Initialize OrdinalEncoder
oe = OrdinalEncoder()

# Fit and transform the categorical data
oe_encoded = oe.fit_transform(df[large_cat_columns].fillna("Unknown"))

# Create a DataFrame for the encoded values and directly substitute into the original df
df[large_cat_columns] = pd.DataFrame(oe_encoded, columns=large_cat_columns, index=df.index)
```

```python
# Plotting the salary distribution (assuming 'Salary' is the correct column)
plt.figure(figsize=(10, 6))
plt.hist(df['converted_comp_yearly'].dropna())
plt.title('Salary Distribution of Users')
plt.xlabel('Salary')
plt.ylabel('Frequency')
plt.show()
```

```python
# Define bins: from 0 to 200k in 20k increments + overflow bin for 200k+
bins = np.arange(0, 220001, 20000)

# Clip salary values, capping them at 200k+
clipped_salaries = np.clip(df['converted_comp_yearly'].dropna(), bins[0], bins[-1])

# Plot the histogram, fig size make a plot of 9x5 inchs
# Create a figure and axes object used to edit the fig and axes (not need fig / ax are low level, plt can do everything)
plt.subplots(figsize=(9, 5))
plt.hist(clipped_salaries, bins=bins, edgecolor = 'black')

# Label the diagram
plt.title('Salary Distribution of Users')
plt.xlabel('Salary')
plt.ylabel('Frequency')

# Format x-axis for better readability
xlabels = bins[0:].astype(int).astype(str)
xlabels[-1] = '200000+'

#limits the range of the x axis to fit the range of our data (remove unnecessary white space)
plt.xlim([0, 220000])

# Position the label of the ticks to the end of each and the corresponding labels
plt.xticks(20000 * np.arange(len(xlabels)), labels = xlabels)

# Show the plot
plt.show()
```

```python
# Filter out salaries greater than 200k
filtered_salaries = df['converted_comp_yearly'].dropna()
filtered_salaries = filtered_salaries[filtered_salaries <= 200000]

# Create a box plot
plt.boxplot(filtered_salaries, vert=False,
           boxprops=dict(color='blue'),
           whiskerprops=dict(color='blue'))

# Label the diagram
plt.title('Salary Distribution of Users')
plt.xlabel('Salary')

# Show the plot
plt.show()
```

# Machine Learning

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression
import pandas as pd

df_cleaned = df.dropna()

continuous_cols = ['main_branch', 'age', 'remote_work', 'ed_level',
                   'years_code', 'years_code_pro', 'dev_type', 'org_size',
                   'purchase_influence', 'buildvs_buy', 'comp_total']

x = df_cleaned[continuous_cols]
y = df_cleaned["converted_comp_yearly"]

X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=69)

lin_reg = LinearRegression()
lin_reg.fit(X_train, y_train)

score = lin_reg.score(X_test, y_test)
print("R² Score:", score)
```
